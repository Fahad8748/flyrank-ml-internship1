```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_score, recall_score, roc_auc_score
from sklearn.model_selection import GroupKFold, train_test_split

# Set seed for reproducibility
np.random.seed(42)

# --- 1. Load / Construct Dataset ---
# Replace with your actual FlyRank dataset loading logic:
# df = pd.read_csv('../data/flyrank_week5_model_data.csv')

n_samples = 1500
clients = [f"client_{i:03d}" for i in range(1, 61)]  # 60 unique clients

df = pd.DataFrame(
    {
        "client_id": np.random.choice(clients, size=n_samples),
        "timestamp": pd.date_range(
            start="2026-01-01", periods=n_samples, freq="H"
        ),
        "session_duration": np.random.exponential(scale=100, size=n_samples),
        "pages_visited": np.random.poisson(lam=4, size=n_samples),
        "target_converted": np.random.choice(
            [0, 1], size=n_samples, p=[0.75, 0.25]
        ),
    }
)

# Introduce a client-level latent effect (common source of leak under random split)
client_conversion_rate = df.groupby("client_id")["target_converted"].transform(
    "mean"
)
df["historical_client_signal"] = client_conversion_rate + np.random.normal(
    0, 0.1, n_samples
)

X = df[
    ["session_duration", "pages_visited", "historical_client_signal"]
].copy()
y = df["target_converted"]
groups = df["client_id"]

# --- 2. Naive Random Split (Over-optimistic / Leaky) ---
X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model_naive = RandomForestClassifier(n_estimators=100, random_state=42)
model_naive.fit(X_train_r, y_train_r)

preds_naive = model_naive.predict_proba(X_test_r)[:, 1]
auc_naive = roc_auc_score(y_test_r, preds_naive)

# --- 3. Honest Grouped Split (Grouped by client_id) ---
gkf = GroupKFold(n_splits=5)
train_idx, test_idx = next(gkf.split(X, y, groups=groups))

X_train_g, X_test_g = X.iloc[train_idx], X.iloc[test_idx]
y_train_g, y_test_g = y.iloc[train_idx], y.iloc[test_idx]

model_honest = RandomForestClassifier(n_estimators=100, random_state=42)
model_honest.fit(X_train_g, y_train_g)

preds_honest = model_honest.predict_proba(X_test_g)[:, 1]
auc_honest = roc_auc_score(y_test_g, preds_honest)

# --- 4. Print Comparison Summary ---
print("==================================================")
print("       FLYRANK MODEL VALIDATION AUDIT SUMMARY     ")
print("==================================================")
print(f"Naive Random Split AUC-ROC  : {auc_naive:.4f}")
print(f"Honest Group-Aware Split AUC: {auc_honest:.4f}")
print(f"Validation Inflation Delta  : {auc_naive - auc_honest:.4f}")
print("==================================================")
```

```python
# --- 1. Correlation Matrix to Identify Target Leakage ---
X_audit = X.copy()
X_audit["TARGET"] = y

plt.figure(figsize=(8, 5))
correlation_matrix = X_audit.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="Blues", fmt=".2f")
plt.title("Feature Leakage Audit - Pearson Correlation Matrix")
plt.tight_layout()
plt.show()

# --- 2. Inspect Error Examples Under Honest Split ---
X_test_g_eval = X_test_g.copy()
X_test_g_eval["actual_target"] = y_test_g
X_test_g_eval["predicted_prob"] = preds_honest
X_test_g_eval["predicted_class"] = (preds_honest >= 0.5).astype(int)

# Identify False Positives and False Negatives
false_positives = X_test_g_eval[
    (X_test_g_eval["actual_target"] == 0)
    & (X_test_g_eval["predicted_class"] == 1)
]
false_negatives = X_test_g_eval[
    (X_test_g_eval["actual_target"] == 1)
    & (X_test_g_eval["predicted_class"] == 0)
]

print(f"\n--- Failure Diagnostics ---")
print(f"Total Test Instances: {len(X_test_g_eval)}")
print(f"False Positives Count: {len(false_positives)}")
print(f"False Negatives Count: {len(false_negatives)}")

print("\nSample False Negative Failure Cases:")
display(false_negatives.head(3))
```
