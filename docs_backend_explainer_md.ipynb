```python
import os

# Ensure directories exist
os.makedirs("public", exist_ok=True)
os.makedirs("docs", exist_ok=True)

# -------------------------------------------------------------------
# STEP 1: Update public/index.html with Live Dynamic Netlify Form
# -------------------------------------------------------------------
updated_index_html = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Muhammad Fahad Rafique | ML Engineer & Bioinformatician</title>
    <style>
        :root {
            --bg-color: #0d1117;
            --card-bg: #161b22;
            --border-color: #30363d;
            --text-main: #c9d1d9;
            --text-heading: #ffffff;
            --accent: #58a6ff;
            --accent-hover: #1f6feb;
            --input-bg: #0d1117;
            --success-color: #2ea043;
        }
        body {
            font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, Helvetica, Arial, sans-serif;
            background-color: var(--bg-color);
            color: var(--text-main);
            margin: 0;
            padding: 0;
            line-height: 1.6;
        }
        .container {
            max-width: 800px;
            margin: 40px auto;
            padding: 0 20px;
        }
        header {
            border-bottom: 1px solid var(--border-color);
            padding-bottom: 24px;
            margin-bottom: 32px;
        }
        h1 {
            color: var(--text-heading);
            font-size: 2.2rem;
            margin: 0 0 8px 0;
        }
        .subtitle {
            font-size: 1.1rem;
            color: var(--accent);
            margin-bottom: 16px;
        }
        .bio {
            font-size: 1rem;
            margin-bottom: 24px;
        }
        .links {
            display: flex;
            gap: 12px;
            flex-wrap: wrap;
            margin-bottom: 32px;
        }
        .btn {
            background-color: var(--card-bg);
            color: var(--accent);
            border: 1px solid var(--border-color);
            padding: 8px 16px;
            border-radius: 6px;
            text-decoration: none;
            font-weight: 500;
            transition: all 0.2s ease;
            cursor: pointer;
        }
        .btn:hover {
            background-color: var(--accent-hover);
            color: #ffffff;
            border-color: var(--accent-hover);
        }
        .btn-submit {
            background-color: #238636;
            color: #ffffff;
            border: 1px solid rgba(240, 246, 252, 0.1);
            width: 100%;
            padding: 12px;
            font-size: 1rem;
            font-weight: 600;
            border-radius: 6px;
            margin-top: 8px;
        }
        .btn-submit:hover {
            background-color: #2ea043;
        }
        .section-title {
            color: var(--text-heading);
            font-size: 1.4rem;
            border-bottom: 1px solid var(--border-color);
            padding-bottom: 8px;
            margin-top: 40px;
            margin-bottom: 20px;
        }
        .card {
            background-color: var(--card-bg);
            border: 1px solid var(--border-color);
            border-radius: 6px;
            padding: 20px;
            margin-bottom: 16px;
        }
        .card h3 {
            margin-top: 0;
            color: var(--text-heading);
        }

        /* Contact Form Styles */
        .form-group {
            margin-bottom: 16px;
            text-align: left;
        }
        .form-group label {
            display: block;
            margin-bottom: 6px;
            color: var(--text-heading);
            font-size: 0.95rem;
            font-weight: 500;
        }
        .form-group input, .form-group textarea {
            width: 100%;
            padding: 10px;
            background-color: var(--input-bg);
            border: 1px solid var(--border-color);
            border-radius: 6px;
            color: var(--text-main);
            font-size: 1rem;
            box-sizing: border-box;
        }
        .form-group input:focus, .form-group textarea:focus {
            outline: none;
            border-color: var(--accent);
        }

        footer {
            margin-top: 60px;
            padding-top: 20px;
            border-top: 1px solid var(--border-color);
            font-size: 0.9rem;
            color: #8b949e;
            text-align: center;
        }
    </style>
</head>
<body>
    <div class="container">
        <header>
            <h1>Muhammad Fahad Rafique</h1>
            <div class="subtitle">ML Systems Engineer | Bioinformatician</div>
            <p class="bio">
                Building scalable machine learning pipelines, search intelligence agents, and computational biology solutions.
                Specializing in search volatility tracking, differential gene analysis, and autonomous workflow design.
            </p>
            <div class="links">
                <a href="https://github.com/Fahad8748" target="_blank" class="btn">GitHub</a>
                <a href="https://linkedin.com/in/" target="_blank" class="btn">LinkedIn</a>
                <a href="https://github.com/Fahad8748/flyrank-ml-internship" target="_blank" class="btn">FlyRank Repository</a>
                <a href="#contact" class="btn">Get in Touch</a>
            </div>
        </header>

        <main>
            <h2 class="section-title">Current Projects & Work</h2>
            <div class="card">
                <h3>SearchGuard-Agent (Capstone Project)</h3>
                <p>An autonomous search intelligence agent built in Python to monitor page-level organic rank decay, run target-leakage audits, and output prioritized content refresh queues.</p>
            </div>
            <div class="card">
                <h3>FlyRank AI ML Internship</h3>
                <p>Developing reproducible ML artifacts, temporal data contracts, intent classification pipelines, and autonomous agent loops.</p>
            </div>

            <!-- DYNAMIC FEATURE: NETLIFY SERVERLESS CONTACT FORM -->
            <h2 class="section-title" id="contact">Get in Touch (Live Contact Form)</h2>
            <div class="card">
                <form name="contact" method="POST" data-netlify="true" netlify-honeypot="bot-field">
                    <!-- Required Netlify hidden inputs -->
                    <input type="hidden" name="form-name" value="contact" />
                    <p style="display:none">
                        <label>Don’t fill this out if you're human: <input name="bot-field" /></label>
                    </p>

                    <div class="form-group">
                        <label for="name">Your Name</label>
                        <input type="text" id="name" name="name" placeholder="e.g. Alex Smith" required />
                    </div>

                    <div class="form-group">
                        <label for="email">Your Email Address</label>
                        <input type="email" id="email" name="email" placeholder="alex@example.com" required />
                    </div>

                    <div class="form-group">
                        <label for="message">Message</label>
                        <textarea id="message" name="message" rows="4" placeholder="Type your message or project inquiry here..." required></textarea>
                    </div>

                    <button type="submit" class="btn btn-submit">Send Message</button>
                </form>
            </div>

            <h2 class="section-title">Program Badge</h2>
            <div class="card" id="badge-placeholder">
                <p><em>Official FlyRank completion badge will be displayed here upon capstone approval.</em></p>
            </div>
        </main>

        <footer>
            <p>&copy; 2026 Muhammad Fahad Rafique. Powered by Netlify Forms (HTTPS Serverless Infrastructure).</p>
        </footer>
    </div>
</body>
</html>
"""

with open("public/index.html", "w", encoding="utf-8") as f:
    f.write(updated_index_html)

print("Updated: public/index.html with Netlify Forms integration")

# -------------------------------------------------------------------
# STEP 2: Create docs/backend_explainer.md (Written Plain-Words Explainer)
# -------------------------------------------------------------------
explainer_md = """# Plain-Words Backend & Dynamic Data Flow Explainer

> **Assignment:** Make It Do Something (PF-06 / Week 6)
> **Author:** Muhammad Fahad Rafique
> **Dynamic Feature Implemented:** Live Serverless Contact Form & Email Pipeline (Netlify Forms)

---

## 1. What is a Backend? (Plain-Words Explanation)

A **frontend** is everything a visitor sees and interacts with directly inside their browser — the text, layout, buttons, and HTML/CSS structure. It runs locally on the user's computer.

A **backend** is the hidden engine that runs on a remote server. While the frontend handles display, the backend handles **action**: listening for requests, processing user input, checking permissions, saving data into a database, or sending emails.

When a user submits a static HTML form without a backend, nothing happens because the browser has nowhere to send the data. A backend provides the destination endpoint that receives, validates, and acts on that information.

---

## 2. What My Dynamic Feature Does

I added a **live, fully wired contact form** to my personal portfolio website.

Instead of requiring visitors to manually copy my email address and open an external email app, visitors can type their name, email, and message directly on my site and hit **"Send Message"**.

Behind the scenes, Netlify's serverless form handler captures the entry, inspects it for spam, logs the submission into my Netlify site dashboard, and forwards a notification directly to my inbox in real time.

---

## 3. How the Data Flows (Step-by-Step)

Here is exactly what happens when someone submits the form on my site:

1.  **User Fills Out Form:** A visitor to `fahad-rafique.netlify.app` navigates to the "Get in Touch" section and fills in the `Name`, `Email Address`, and `Message` fields.

2.  **"Send Message" Button Click:** The user clicks the `Submit` button.

3.  **Netlify's HTML Attributes Trigger:** Because the `<form>` tag has `data-netlify="true"` and `name="contact"` attributes, Netlify automatically intercepts the form submission. It doesn't need custom JavaScript or a server-side endpoint I built.

4.  **Submission to Netlify:** The browser sends a POST request containing the form data directly to Netlify's backend services.

5.  **Spam Filtering (Honeypot):** Netlify first checks the `netlify-honeypot="bot-field"` field. This is a hidden field (`style="display:none"`) that human users won't see or fill. If a bot fills it, Netlify knows it's spam and discards the submission silently.

6.  **Data Storage:** If not spam, Netlify stores the submission securely in my site's admin panel under "Forms" -> "Contact" -> "Submissions". This provides a dashboard where I can view all inquiries.

7.  **Email Notification:** Netlify's pre-configured serverless function triggers an email notification, sending the form's content (name, email, message) to my specified email address. This happens in real-time, so I get immediate alerts for new messages.

8.  **Success Page (Default or Custom):** By default, Netlify redirects the user to a generic success page. However, I can configure a custom success page if I want a more branded "Thank You" experience.

---

## 4. Backend Benefits for this Project

*   **No Server Maintenance:** I don't need to spin up a server (like Node.js, Python Flask, or PHP) to handle form submissions. Netlify provides this "serverless" functionality out of the box.
*   **Scalability:** Netlify's forms scale automatically. Whether I get one submission or a thousand, the backend infrastructure handles it without me needing to worry about server load.
*   **Ease of Use:** Integrating the form was as simple as adding a few HTML attributes. No complex backend coding was required.
*   **Reliability:** Netlify's global CDN and robust infrastructure ensure high availability for the form submission endpoint.
*   **Spam Protection:** The honeypot field provides a basic, yet effective, layer of spam protection without needing CAPTCHAs.

This simple form demonstrates how external, managed backend services (like Netlify Forms) can add powerful dynamic functionality to a static frontend with minimal development effort, making the site "do something" meaningful without building a full custom backend.
"""

with open("docs/backend_explainer.md", "w", encoding="utf-8") as f:
    f.write(explainer_md)

print("Created: docs/backend_explainer.md")
```
