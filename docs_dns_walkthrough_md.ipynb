```python
import os

# -------------------------------------------------------------------
# STEP 1: Directory Setup
# -------------------------------------------------------------------
os.makedirs("public", exist_ok=True)
os.makedirs("docs", exist_ok=True)

# -------------------------------------------------------------------
# STEP 2: Create public/index.html (Clean Personal Portfolio)
# -------------------------------------------------------------------
index_html = """<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Fahad | ML Engineer & Bioinformatician</title>
    <style>
        :root {
            --bg-color: #0d1117;
            --card-bg: #161b22;
            --border-color: #30363d;
            --text-main: #c9d1d9;
            --text-heading: #ffffff;
            --accent: #58a6ff;
            --accent-hover: #1f6feb;
        }
        body {
            font-family: -apple-system, BlinkMacSystemFont, "Segoe UI", Roboto, Helvetica, Arial, sans-serif;
            background-color: var(--bg-color);
            color: var(--text-main);
            margin: 0;
            padding: 0;
            line-height: 1.6;
        }
        .container {
            max-width: 800px;
            margin: 40px auto;
            padding: 0 20px;
        }
        header {
            border-bottom: 1px solid var(--border-color);
            padding-bottom: 24px;
            margin-bottom: 32px;
        }
        h1 {
            color: var(--text-heading);
            font-size: 2.2rem;
            margin: 0 0 8px 0;
        }
        .subtitle {
            font-size: 1.1rem;
            color: var(--accent);
            margin-bottom: 16px;
        }
        .bio {
            font-size: 1rem;
            margin-bottom: 24px;
        }
        .links {
            display: flex;
            gap: 12px;
            flex-wrap: wrap;
            margin-bottom: 32px;
        }
        .btn {
            background-color: var(--card-bg);
            color: var(--accent);
            border: 1px solid var(--border-color);
            padding: 8px 16px;
            border-radius: 6px;
            text-decoration: none;
            font-weight: 500;
            transition: all 0.2s ease;
        }
        .btn:hover {
            background-color: var(--accent-hover);
            color: #ffffff;
            border-color: var(--accent-hover);
        }
        .section-title {
            color: var(--text-heading);
            font-size: 1.4rem;
            border-bottom: 1px solid var(--border-color);
            padding-bottom: 8px;
            margin-top: 40px;
            margin-bottom: 20px;
        }
        .card {
            background-color: var(--card-bg);
            border: 1px solid var(--border-color);
            border-radius: 6px;
            padding: 20px;
            margin-bottom: 16px;
        }
        .card h3 {
            margin-top: 0;
            color: var(--text-heading);
        }
        footer {
            margin-top: 60px;
            padding-top: 20px;
            border-top: 1px solid var(--border-color);
            font-size: 0.9rem;
            color: #8b949e;
            text-align: center;
        }
    </style>
</head>
<body>
    <div class="container">
        <header>
            <h1>Muhammad Fahad Rafique</h1>
            <div class="subtitle">ML Systems Engineer | Bioinformatician</div>
            <p class="bio">
                Building scalable machine learning pipelines, search intelligence agents, and computational biology solutions.
                Specializing in search volatility tracking, differential gene analysis, and autonomous workflow design.
            </p>
            <div class="links">
                <a href="https://github.com/Fahad8748" target="_blank" class="btn">GitHub</a>
                <a href="https://linkedin.com/in/" target="_blank" class="btn">LinkedIn</a>
                <a href="https://github.com/Fahad8748/flyrank-ml-internship" target="_blank" class="btn">FlyRank Repository</a>
                <a href="mailto:contact@example.com" class="btn">Book a Call / Email</a>
            </div>
        </header>

        <main>
            <h2 class="section-title">Current Projects & Work</h2>
            <div class="card">
                <h3>SearchGuard-Agent (Capstone Project)</h3>
                <p>An autonomous search intelligence agent built in Python to monitor page-level organic rank decay, run target-leakage audits, and output prioritized content refresh queues.</p>
            </div>
            <div class="card">
                <h3>FlyRank AI ML Internship</h3>
                <p>Developing reproducible ML artifacts, temporal data contracts, intent classification pipelines, and autonomous agent loops.</p>
            </div>

            <h2 class="section-title">Program Badge</h2>
            <div class="card" id="badge-placeholder">
                <p><em>Official FlyRank completion badge will be displayed here upon capstone approval.</em></p>
            </div>
        </main>

        <footer>
            <p>&copy; 2026 Muhammad Fahad Rafique. Hosted on Netlify over HTTPS.</p>
        </footer>
    </div>
</body>
</html>
"""

with open("public/index.html", "w", encoding="utf-8") as f:
    f.write(index_html)

print("Created: public/index.html")

# -------------------------------------------------------------------
# STEP 3: Create docs/dns_walkthrough.md (Written DNS Deliverable)
# -------------------------------------------------------------------
dns_walkthrough_md = """# DNS & Infrastructure Walkthrough (PF-04 Deliverable)

> **Author:** Muhammad Fahad Rafique
> **Topic:** How the Domain Name System (DNS) maps human-readable web addresses to servers.

---

## 1. What is DNS? (The Internet's Phonebook)
Computers communicate across the internet using numerical IP addresses (such as `192.0.2.1` or `2606:4700::6811:d509`). Because strings like `fahad-rafique.netlify.app` are easier for human beings to read and remember than raw IP strings, the **Domain Name System (DNS)** translates human-friendly web addresses into machine-readable IP addresses.

---

## 2. What Happens When Someone Types a Website Address?

When a user opens a web browser and types `https://fahad-rafique.netlify.app`, a step-by-step query process occurs in milliseconds:

1. **Browser & OS Cache:** The user's device first checks its internal memory (cache) to see if it already knows the IP address for `fahad-rafique.netlify.app`. If missing, it sends a request out to the internet.
2. **DNS Recursive Resolver:** The request arrives at the **Resolver** (usually operated by your internet provider or a public service like 1.1.1.1). The resolver acts as a middleman tasked with finding the correct IP.
3. **Root Nameserver (`.`):** The resolver queries a Root Server, which points the resolver toward the servers handling `.app` top-level domains.
4. **TLD Nameserver (`.app`):** The Top-Level Domain (TLD) server reads the request for `.app` and directs the resolver to Netlify's authoritative nameservers.
5. **Authoritative Nameserver:** Netlify's DNS server holds the official DNS records. It checks the host record for `fahad-rafique` and responds with the exact IP address of the web server storing the site files.
6. **Browser Connection over HTTPS:** The resolver passes this IP back to your browser. The browser contacts Netlify's host server over port 443 (HTTPS), negotiates a secure SSL/TLS encryption handshake, downloads `index.html`, and renders the site.

---

## 3. What is a CNAME Record?

A **CNAME (Canonical Name) Record** is an alias entry in DNS. Instead of mapping a domain directly to an IP address, a CNAME record points one domain name to another domain name.

* **Example Use Case:** If you register a custom domain like `fahad-ml.com` and want `www.fahad-ml.com` to show your Netlify site, you create a CNAME record:
  * **Name:** `www.fahad-ml.com`
  * **Type:** `CNAME`
  * **Value/Target:** `fahad-rafique.netlify.app`

When someone visits `www.fahad-ml.com`, the DNS resolver looks up the CNAME record, sees that it points to `fahad-rafique.netlify.app`, resolves that host address, and serves the content. Using a CNAME allows hosting providers like Netlify to update host server IP addresses behind the scenes without breaking your custom domain settings.

---

## 4. Deployed Files Explanation

* `public/index.html`: The complete single-page website containing semantic HTML structure, responsive CSS styling, portfolio links (GitHub, LinkedIn, repository), and a designated slot for the upcoming FlyRank completion badge.
"""

with open("docs/dns_walkthrough.md", "w", encoding="utf-8") as f:
    f.write(dns_walkthrough_md)

print("Created: docs/dns_walkthrough.md")
print("\nSUCCESS: All files for Assignment PF-04 generated successfully!")
```
